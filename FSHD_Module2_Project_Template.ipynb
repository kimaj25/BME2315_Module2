## Project Goal:
Facioscapulohumeral muscular dystrophy (FSHD) is a progressive neuromuscular disorder characterized by skeletal muscle weakness, degeneration, and the gradual replacement of muscle tissue with fat. As the disease progresses, patients may experience increasing difficulty performing everyday activities, including walking, standing, and maintaining balance. However, FSHD progression varies considerably between individuals and often occurs gradually, making it challenging to detect meaningful changes in disease severity over relatively short periods. This presents a significant challenge in clinical trials, where researchers must identify reliable and sensitive outcome measures to evaluate the effectiveness of potential treatments.

The goal of this project is to investigate whether MRI-derived muscle fat fraction (FF) and Timed Up and Go (TUG) performance are correlated in patients with FSHD, and to evaluate their potential usefulness as indicators of disease progression in clinical trials. Muscle fat fraction is a quantitative imaging biomarker that measures the proportion of fat within skeletal muscle, reflecting the extent of fatty replacement associated with muscle degeneration. In contrast, the TUG test is a functional assessment that measures the time required for a patient to stand from a seated position, walk a short distance, turn around, return, and sit down. Longer completion times generally indicate reduced functional mobility. While fat fraction measures structural changes within the muscle, TUG time provides a broader assessment of a patient's physical function.

We propose the questions: 

1. Does right-leg fat fraction or TUG change over common follow up intervals?

2. Are higher fat fractions associated with longer TUG times?

3. Do changes in fat fraction accompany changes in TUG?


To investigate this relationship, we will analyze longitudinal natural-history data collected from FSHD patients at baseline, year 1, and year 2. Our analysis will examine the relationship between fat fraction and TUG time at each available timepoint using scatterplots, correlation coefficients, and linear regression. We will also evaluate changes in both measurements over time to determine whether increases in muscle fat fraction correspond with increases in TUG completion time within individual patients. Statistical measures, including correlation coefficients, R², and p-values, will be used to characterize the strength and statistical significance of these relationships.

Ultimately, this project aims to determine whether structural muscle degeneration, as quantified through MRI-derived fat fraction, is associated with measurable differences in functional mobility. Understanding this relationship is important for evaluating the usefulness of fat fraction and TUG time as indicators in FSHD clinical trials. If the measurements demonstrate a strong relationship, they may provide complementary evidence of disease progression by connecting structural muscle changes with functional impairment. Conversely, a weak association may suggest that the two measurements capture distinct aspects of the disease. By examining their relationship and longitudinal changes, we will assess whether both measures should be considered when monitoring disease progression and evaluating potential therapeutic interventions.


## Disease Background: 

* **Prevalence & incidence**
   * The prevalence of FSHD is around four to ten people per 100,000 ([Preston & Wang, 2025](https://www.ncbi.nlm.nih.gov/books/NBK1443/)). There are differences across populations, with reports out of the Netherlands being as high as twelve per 100,00 ([Deenen et al., 2014](https://pmc.ncbi.nlm.nih.gov/articles/PMC4166358/?utm_source=chatgpt.com)) and as low as one per 435,000 in Wisconsin in the United States. Globally this accounts for around 870,000 people worldwide ([FSHD Society, 2019](https://www.fshdsociety.org/living-with-fshd/understanding-fshd/)). Although many cases are linked to genetics, around 30% of cases arise without prior family history. Incidence of FSHD is about 1 in 7500 births ([UMass Chan, 2014](https://www.umassmed.edu/wellstone/overview/fshdfacts/)).

* **Symptoms**
   * Common symptoms of FSHD include facial weakness (one of the earliest signs), shoulder weakness (lifting arms, throwing objects, reaching), lower leg weakness (foot drop), abdominal weakness (posture changes), hip weakness (difficulty standing, climbing stairs, walking, etc.), and general total-body pain and inflammation. 
   * Other characteristic developments of FSHD that appear less commonly/later in the disease’s progression are spinal/joint alignment changes, eye changes (abnormal blood vessels causing blindness), hearing loss, and dysphagia ([Muscular Dystrophy Association, 2015](https://www.mda.org/disease/facioscapulohumeral-muscular-dystrophy/signs-and-symptoms)).
   
   Broader Lens:
   * Though not a direct physiological effect of the disease, FSHD patients are shown to endure mental health challenges. This was highlighted in Claire Anderson’s talk about FSHD awareness, where she revealed her own struggles with depression and suicidal thoughts as a result of the hopelessness she felt in her disabled body ([Muscular Dystrophy Association, 2015](https://www.youtube.com/watch?v=dDAViLzpms8)). In a study with 700 participants, about 1 out of 4 participants self-reported a diagnosis of depression (28.5%); a nearly identical number (28.3%) self-reported a diagnosis of anxiety, and 28% of participants are currently taking antidepressant medication ([Sasala, 2026a](https://www.fshdsociety.org/2026/05/22/betterlife-mental-health-glimpse/)).

* **Diagnosis**
   * Diagnosis of FSHD often begins with patients noticing unusual weakness during exercise or everyday movement. Often genetic testing is performed to confirm FSHD diagnosis. For FSHD1, testing on the D4Z4 repeating region on the fourth chromosome is performed with FSHD1 patients having limited repetitions of D4Z4 (under ten). Patients with FSHD2 have variants for the SCHMD1 gene which interacts with the fourth chromosome. Testing for FSHD2 is performed when testing for FSHD1 is negative ([BlueCross BlueShield Tennessee, 2024](https://www.bcbst.com/mpmanual/!ssl!/webhelp/Genetic_Testing_for_Facioscapulohumeral_Muscular_Dystrophy.htm)). Both limited D4Z4 and activation of SCHMD1 can cause the activation of the DUX4 gene causing FSHD. Around 95% of affected men will show outward signs of FSHD by the second decade of their lives while 95% of women typically show outward signs by the third decade of their lives ([UMass Chan, 2014](https://www.umassmed.edu/wellstone/overview/fshdfacts/)). 

* **Standard of care treatments (& reimbursement)**
   * Currently, there is no definitive cure or medicine for FSHD that can slow or reverse its progression. Instead, the standard of care relies on proactively managing symptoms and maintaining function. Current therapies include physical and occupational therapy (maintain muscle strength, joint flexibility, and balance), exercise, orthopedic intervention (medical/surgical procedures to target, support, and preserve muscles and joints), pain management via NSAIDs, respiratory surveillance (monitoring sleep), eye care (retinal exam, photocoagulation, patches, and hydrant gels), and hearing loss monitoring ([Aguirre et al., 2023](https://pmc.ncbi.nlm.nih.gov/articles/PMC10316996/)).
   * Since there is no direct solution to FSHD, insurance reimbursement is scattered across various clinical and therapy services.The FSHD Society collaborated with researchers studying insurance claims from Medicare and commercial health plans. The study found that the FSHD patients averaged a yearly insurance claim estimate of $19,400 for FSHD patients with private insurance. In comparison, people without FSHD averaged only about $5,250 in yearly insurance claims. Those with PSHD also had higher prescription drug costs. 
   * Insurance claims only capture the services billed to a health plan, and do not consider the other expenses dealt with by the families involved. Other expenses include unpaid care from family members, lost wages, home modifications, transportation costs, and other out-of-pocket expenses. The numbers from the study also only cover around 4 years of medical claims, rather than a lifetime impact of the progressive disease (indicating the numbers would be even higher).
   * Furthermore, reimbursement (by insurance) rarely covers the total charge of said treatments, causing patient care to be delayed or force the medical service/provider to eat the financial loss or redirect the bill to the patient ([Sasala, 2026b](https://www.fshdsociety.org/2026/08/13/new-study-shows-fshd-has-a-high-cost-and-the-real-cost-is-likely-even-higher/)). 


* **Disease progression & prognosis**
   * The tricky aspect regarding FSHD is the disparity of onset, severity, and progression of the disease among patients. FSHD typically begins with weakening of muscles in the face, arms, scapula, and shoulders, usually presenting asymmetrically. As the disease progresses, weakness spreads down the body to the legs, resulting in around 20% of FSHD patients requiring a wheelchair by age 50 ([FSHD Society, 2019](https://www.fshdsociety.org/living-with-fshd/understanding-fshd/)). FSHD typically does not slow down over time with a five year study concluding a statistically significant result of progression of general muscle weakness in a 154 sample size group with a median age of 51.4 ([Kools et al., 2025](https://pmc.ncbi.nlm.nih.gov/articles/PMC11632561/?utm_source=chatgpt.com)). Although FSHD progresses slowly over time, it does not stop with age.

* **Biological mechanisms (anatomy, organ physiology, cell & molecular physiology)**
   * FSHD is centered on the expression of the double homeobox 4 transcription factor–otherwise known as the DUX4 gene–in skeletal muscle. The genetic alteration manifested in FSHD begins on a specific region of the chromosome 4q35, where the repetitive element (11-100 Kb-sized) tandem repeats of the D4Z4 unit is located within heterochromatin. When someone has at least one copy of the DUX4 gene, they will carry a D4Z4 with a reduced number of 1-10 repeats, known as a contraction. This contraction of D4Z4 repeats results in hypomethylation and a decrease in the repressive heterochromatin of the D4Z4 repeats (referred to as chromatin relaxation). Hence, DUX4 is able to be transcribed, when it normally is not. This mechanism then leads to the muscle atrophy and fatty infiltration seen in FSHD patients ([Hamel & Tawil, 2018](https://pmc.ncbi.nlm.nih.gov/articles/PMC6277282/)).


* **Clinical Trials/next-gen therapies**
   * Current next-gen therapy for patients with FSHD consists of targeting the DUX4 gene to stop its expression, thus hopefully stopping the progression of FSHD. Current trials have tested losmapimod on regulating DUX4 expression, although no statistical significance has been found ([Tawil et al., 2024](https://pubmed.ncbi.nlm.nih.gov/38631764/)). Next-gen therapies have shown interest in using epigenetic editing therapy to repress expression of DUX4. This study is still underway and conclusions about the effectiveness and tolerability of gene editing to treat FSHD have not been found ([University of California Health, 2025](https://clinicaltrials.ucbraid.org/trial/NCT06907875?utm_source=chatgpt.com)).



## Data Set:
The study cohort represented in the dataset included 34 FSHD subjects, 16 female and 18 male, aged 21 to 69 (mean 47.1 ± 14SD) recruited at three sites (University of Washington Medical Center, University of Rochester Medical Center, and Kansas University Medical Center) of the Seattle Paul D. Wellstone Muscular Dystrophy Specialized Research Center. 

The project focuses on the tibialis anterior (TA), a lower-leg muscle responsible for ankle dorsiflexion that can be affected by FSHD. The dataset combines quantitative MRI-derived measurements with genetic biomarkers, clinical disease severity scores, muscle strength Demographic and genetic information includes subject identification, sex, age at baseline, D4Z4 repeat count, and DUX4 gene-expression signatures measured from the right and left sides at baseline. The D4Z4 repeat count represents the number of repeated DNA sequences within the D4Z4 region, while the DUX4 gene-expression signature measures the expression of genes associated with DUX4 activity.
The dataset also includes MRI-derived measurements of muscle fat fraction (FF), fat fraction variance (FVAR), and quantitative short tau inversion recovery (STIR) signals. Fat fraction represents the proportion of fat within the measured muscle tissue, while fat fraction variance describes the variability in fat distribution. STIR measurements quantify water-sensitive MRI signal abnormalities within the muscle. These measurements are recorded separately for the right and left sides, with fat fraction and fat fraction variance available at baseline, year 1, and year 2, and STIR measurements available at baseline and year 1.

Clinical and functional measurements include bilateral ankle dorsiflexion strength assessed through quantitative muscle testing (QMT), the Timed Up and Go (TUG) test, the 10-meter walk/run test, and timed four-stair ascent and descent tests. Dorsiflexion QMT measures the strength of the muscles responsible for lifting the foot upward, while the functional tests record the time required to complete specific movements involving walking, standing, and stair navigation. Additionally, clinical severity scores provide a standardized assessment of disease involvement. These clinical and functional measurements are recorded at baseline, year 1, and year 2. The dataset also includes the elapsed time in months between baseline and each follow-up visit to account for differences in the timing of patient assessments. Note that not every subject has complete data at every visit. 

The tibialis anterior (TA) MRI data were processed using T1-weighted Dixon scans, which separates the magnetic resonance signal from water and fat. This produces two image volumes: a water-image file that contains the MRI signal associated primarily with water-containing tissue, including skeletal muscle; and a fat-image file that contains the MRI signal associated primarily with fat, allowing fatty infiltration of muscle to be evaluated. The MRI scans were processed using a combination of artificial intelligence (AI)-based segmentation and manual verification by Springbok Analytics (Charlottesville, VA). The AI algorithm initially segmented the tibialis anterior muscle across all axial MRI slices. These segmentations were subsequently reviewed by trained segmentation engineers and evaluated by a researcher to ensure consistency, producing a final three-dimensional (3D) segmentation mask of the TA muscle. The mask identifies the voxels corresponding to the tibialis anterior across the MRI slices, allowing muscle volume and fat infiltration to be calculated specifically within the segmented region.

The water and fat images represent the same anatomical region and can be combined to calculate a fat fraction, which measures the ratio of the signal coming from mobile fat protons to the total combined signal from both fat and water protons in the tissue. These intensities are generally expressed in arbitrary signal units and are dimensionless, conventionally expressed as a percentage ([Bray et al., 2017](https://pubmed.ncbi.nlm.nih.gov/28936896/)). 
A segmentation mask identifies which image voxels belong to the tibialis anterior muscle. It separates the muscle from surrounding bones, subcutaneous fat, and neighboring muscles. The mask is a spatial label map rather than an additional MRI measurement. Each voxel has a label indicating whether it belongs to the selected anatomical structure. When applied to the aligned water and fat images, the mask allows quantitative analysis to be restricted to the tibialis anterior. 

Muscle biopsy samples were collected from the tibialis anterior (TA) to evaluate skeletal muscle gene expression, including DUX4-associated gene activity. However, some biopsies contained insufficient muscle tissue due to fatty replacement, blood contamination, or sampling of fibrotic regions. To assess sample quality, researchers measured the relative representation of skeletal muscle, fat, and blood using gene-expression scores calculated from scaled transcripts per million. Two right-sided biopsy samples, 13-0009R and 13-0007R, exhibited particularly low muscle content and elevated fat content. Principal component analysis further supported their low muscle representation. Consequently, these samples were excluded from analyses comparing RNA expression, including DUX4 signatures, with MRI characteristics and clinical measurements. This quality-control process helped ensure that the gene-expression data used in these comparisons adequately represented skeletal muscle tissue ([Wong et al., 2024](https://academic.oup.com/hmg/article/33/8/698/7588694)).

The supplied mask helped identify which pixels belonged the the tibialis anterior.
```python
row = manifest[manifest["timepoint"] == "TP1"].iloc[3]  # representative mid-muscle slice

water = np.load(DATA_DIR / row["water_file"])
fat = np.load(DATA_DIR / row["fat_file"])
mask = np.load(DATA_DIR / row["mask_file"]).astype(bool)

print("Water image shape:", water.shape)
print("Fat image shape:", fat.shape)
print("Mask shape:", mask.shape)
print("Number of muscle pixels:", mask.sum())
#Next code was written to Visualize the water, fat, and mask images and pair 
# them with their respective titles. The images are displayed in a 1x3 grid using Matplotlib's subplots, 
# and the axes are turned off for a cleaner presentation. The water image is shown in grayscale, 
# the fat image is also in grayscale, and the mask is displayed with a binary colormap to highlight the muscle region.
```

## Fat Fraction
Fat fraction was calculated at each pixel. Two types of pixels were considered: pixels in the image region and pixels within the TA mask.

```python
denominator = water + fat
valid_pixels = denominator > 0

fat_fraction = np.divide(
    fat,
    denominator,
    out=np.zeros_like(fat, dtype=float),
    where=valid_pixels
) * 100

# Keep the whole supplied image region for the first analysis.
fat_fraction_full = fat_fraction.copy()

# Then apply the TA mask for the muscle-specific analysis.
fat_fraction_masked = fat_fraction.copy()
fat_fraction_masked[~mask] = np.nan

whole_image_values = fat_fraction_full[valid_pixels]
muscle_values = fat_fraction_masked[~np.isnan(fat_fraction_masked)]

print(f"Whole supplied image region — mean: {np.mean(whole_image_values):.1f}%")
print(f"Whole supplied image region — minimum: {np.min(whole_image_values):.1f}%")
print(f"Whole supplied image region — maximum: {np.max(whole_image_values):.1f}%")
print()
print(f"TA muscle only — mean: {np.mean(muscle_values):.1f}%")
print(f"TA muscle only — minimum: {np.min(muscle_values):.1f}%")
print(f"TA muscle only — maximum: {np.max(muscle_values):.1f}%")

#Using the data from the CSV files, the code calculates the 
#fat fraction for the entire TA slice and muscle region.
```

## Summarizing Distribution
Images of the muscle region were created as: water image + TA mask, fat image + TA mask, fat fraction: whole region, and TA fat fraction. A histogram was also created to show the distrubtion of fat fraction values inside of the TA mask. The horizontal axis represented fat fraction in percent, and the vertical axis represented pixel count. These distributions showed the spread of measurements.
```python
def calculate_fat_fraction(water, fat, mask):
    """Return a masked fat-fraction image in percent."""
    denominator = water + fat
    result = np.divide(
        fat,
        denominator,
        out=np.zeros_like(fat, dtype=float),
        where=denominator > 0
    ) * 100
    result[~mask] = np.nan
    return result

analysis_timepoint = "TP1"  # analyze one visit at a time
analysis_manifest = manifest[manifest["timepoint"] == analysis_timepoint]
results = []

for _, row in analysis_manifest.iterrows():
    water = np.load(DATA_DIR / row["water_file"])
    fat = np.load(DATA_DIR / row["fat_file"])
    mask = np.load(DATA_DIR / row["mask_file"]).astype(bool)
    ff = calculate_fat_fraction(water, fat, mask)

    results.append({
        "subject": row["subject"],
        "timepoint": row["timepoint"],
        "slice_id": row["slice_id"],
        "percent_along_muscle": row["percent_along_muscle"],
        "distance_from_muscle_start_mm": row["distance_from_muscle_start_mm"],
        "mean_fat_fraction_percent": np.nanmean(ff),
        "median_fat_fraction_percent": np.nanmedian(ff),
        "std_fat_fraction_percent": np.nanstd(ff),
        "n_muscle_pixels": int(mask.sum()),
    })

results_df = pd.DataFrame(results)
results_df
# This code calculates the fat fraction for each slice in the manifest for a 
# specific timepoint (TP1). It iterates through each row of the filtered manifest, 
# loads the corresponding water, fat, and mask images, and computes the fat fraction using the `calculate_fat_fraction` function. 
# The results, including subject information and statistical measures of the fat fraction, 
# are stored in a list of dictionaries and then converted into a DataFrame for further analysis or visualization.
```

```python
def summarize_visit(timepoint):
    visit_results = []
    visit_manifest = manifest[manifest["timepoint"] == timepoint]

    for _, row in visit_manifest.iterrows():
        water = np.load(DATA_DIR / row["water_file"])
        fat = np.load(DATA_DIR / row["fat_file"])
        mask = np.load(DATA_DIR / row["mask_file"]).astype(bool)
        ff = calculate_fat_fraction(water, fat, mask)

        visit_results.append({
            "timepoint": timepoint,
            "slice_id": row["slice_id"],
            "percent_along_muscle": row["percent_along_muscle"],
            "distance_from_muscle_start_mm": row["distance_from_muscle_start_mm"],
            "mean_fat_fraction_percent": np.nanmean(ff),
        })

    return pd.DataFrame(visit_results)

profiles_df = pd.concat(
    [summarize_visit("TP1"), summarize_visit("TP2")],
    ignore_index=True
)

plt.figure(figsize=(7, 4))
for timepoint, group in profiles_df.groupby("timepoint"):
    group = group.sort_values("distance_from_muscle_start_mm")
    plt.plot(
        group["percent_along_muscle"],
        group["mean_fat_fraction_percent"],
        marker="o",
        label=timepoint
    )
plt.xlabel("Distance from muscle start (mm)")
plt.ylabel("Mean TA fat fraction (%)")
plt.title("Mean fat fraction along the tibialis anterior")
plt.legend(title="Visit")
plt.grid(True, alpha=0.3)
plt.show()
#This code generates a line plot to visualize the mean fat fraction 
#along the tibialis anterior muscle for 
#two different timepoints (TP1 and TP2).
```

```python
#Import libraries
from pathlib import Path
import argparse
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

#The code below is adapted from the original BME2315 Module 2 analysis script, 
# with modifications for clarity and reproducibility.
parser = argparse.ArgumentParser()
parser.add_argument('csv', nargs='?', default=None)
args, _ = parser.parse_known_args()
HERE = Path(__file__).resolve().parent if '__file__' in globals() else Path.cwd()
CSV = (HERE / 'C:\\Users\\aerin\\OneDrive\\Desktop\\COMP BME 2315 Fall 2026\\Module 2\\BME2315_Module2\\Module2_FSHD_Dataset.csv') if args.csv is None else Path(args.csv)
OUT = HERE / 'fshd_presentation_results'
OUT.mkdir(exist_ok=True)
plt.rcParams.update({'font.size': 12, 'axes.titlesize': 15, 'axes.labelsize': 12})
VISITS = ['baseline', 'y1', 'y2']
TC = [f'lag_months_{v}' for v in VISITS]
COLS = {'FF': [f'ff_{v}_R' for v in VISITS],
        'TUG': [f'tug_time_{v}' for v in VISITS]}
TARGETS = [0., 11.15, 22.13]
LABELS = ['Baseline', 'T1', 'T2']
AXLABEL = {'FF': 'Right-leg fat fraction (%)', 'TUG': 'TUG time (s)'}
UNITS = {'FF': 'percentage points', 'TUG': 's'}

#1. Helper functions for formatting and saving figures
def pf(p):
    return f'{p:.2e}' if p < .001 else f'{p:.4f}'
#This function saves figures in both PNG and PDF formats, 
# and appends the figure caption to a text file for reference.
def save(fig, name, caption):
    fig.savefig(OUT / f'{name}.png', dpi=240, bbox_inches='tight')
    fig.savefig(OUT / f'{name}.pdf', bbox_inches='tight')
    with (OUT / 'figure_captions.txt').open('a') as f:
        f.write(f'{name}: {caption}\n\n')
    plt.show()
    plt.close(fig)
#Add a note to a given axis, turning off the axis and placing the text at the top.
def note(ax, text):
    ax.axis('off')
    ax.text(0, 1, text, va='top', fontsize=12, linespacing=1.55)

(OUT / 'figure_captions.txt').write_text('')

#2. Load, describe, and flag timing issues
raw = pd.read_csv(CSV)
if raw.subject_id.isna().any() or raw.subject_id.duplicated().any():
    raise ValueError('Missing or duplicate subject identifiers')
for col in TC + COLS['FF'] + COLS['TUG']:
    raw[col] = pd.to_numeric(raw[col], errors='raise')
print('Original cohort:', len(raw))
print(raw.sex.value_counts())
print(raw.age_baseline.agg(['min', 'max', 'mean', 'median']))
ambiguous = raw[TC].apply(lambda r: r.dropna().duplicated().any(), axis=1)
raw.loc[ambiguous].to_csv(OUT / 'excluded_ambiguous_times.csv', index=False)
df = raw.loc[~ambiguous].copy()

#3. Linear interpolation and shared cohort

def observed(row, measure):
    t = row[TC].to_numpy(float)
    y = row[COLS[measure]].to_numpy(float)
    k = np.isfinite(t) & np.isfinite(y)
    t, y = t[k], y[k]
    order = np.argsort(t)
    t, y = t[order], y[order]
    if len(t) > 1 and np.any(np.diff(t) <= 0):
        raise ValueError(f'Ambiguous times: {row.subject_id}')
    return t, y

aligned = pd.DataFrame({'subject_id': df.subject_id.to_numpy()})
for measure in COLS:
    values = []
    for _, row in df.iterrows():
        t, y = observed(row, measure)
        est = np.full(3, np.nan)
        for j, target in enumerate(TARGETS):
            exact = np.flatnonzero(np.isclose(t, target, rtol=0, atol=1e-8))
            if len(exact):
                est[j] = y[exact[0]]
            elif len(t) >= 2 and t[0] < target < t[-1]:
                est[j] = np.interp(target, t, y)
        values.append(est)
    for j, label in enumerate(LABELS):
        aligned[f'{measure}_{label}'] = np.asarray(values)[:, j]
needed = [f'{m}_{l}' for m in COLS for l in LABELS]
aligned['included'] = aligned[needed].notna().all(axis=1)
cohort = aligned.loc[aligned.included].copy()
aligned.to_csv(OUT / 'aligned_values_and_inclusion.csv', index=False)
print('Shared cohort:', len(cohort))
if len(cohort) < 3:
    raise ValueError('Fewer than three subjects have all aligned values. Adjust target times or inclusion criteria.')

#4. Verify interpolation with measured points and estimates
sid = cohort.subject_id.iloc[0]
row = df.loc[df.subject_id == sid].iloc[0]
est = cohort.loc[cohort.subject_id == sid].iloc[0]
fig, axes = plt.subplots(1, 2, figsize=(12, 4.8))
for ax, m in zip(axes, COLS):
    t, y = observed(row, m)
    ax.plot(t, y, 'o-', label='Observed measurements')
    yy = [est[f'{m}_{l}'] for l in LABELS]
    ax.scatter(TARGETS, yy, marker='x', s=100, color='darkorange',
               label='Common-time values', zorder=3)
    for x, val in zip(TARGETS, yy):
        ax.annotate(f'{val:.2f}', (x, val), xytext=(5, 9), textcoords='offset points')
    ax.set(xlabel='Months since baseline', ylabel=AXLABEL[m], title=f'Subject {sid}: {m}')
    ax.legend(fontsize=9)
fig.suptitle('Linear interpolation: 0, 11.15, and 22.13 months')
fig.tight_layout()
save(fig, '01_interpolation_example', 'Crosses are estimated common-time values; circles are original measurements. Straight segments assume linear change;' \
' no extrapolation is used.')

#5. Paired tests, confidence intervals, changes, and Holm adjustment
# This section computes paired t-tests for each measure across the three intervals, 
# calculates mean changes and confidence intervals, and applies Holm's method for multiple comparisons.
rows = []
for start, end, interval in [('Baseline', 'T1', '0 to 11.15 months'),
                             ('T1', 'T2', '11.15 to 22.13 months'),
                             ('Baseline', 'T2', '0 to 22.13 months')]:
    for m in COLS:
        a, b = cohort[f'{m}_{start}'], cohort[f'{m}_{end}']
        d = b - a
        pc = 100 * d / a.where(a != 0)
        n = len(d)
        margin = stats.t.ppf(.975, n-1) * stats.sem(d)
        test = stats.ttest_rel(b, a)
        rows.append(dict(measure=m, interval=interval, n=n,
            mean_start=a.mean(), mean_end=b.mean(), mean_change=d.mean(),
            sd_change=d.std(ddof=1), ci_low=d.mean()-margin, ci_high=d.mean()+margin,
            mean_pct=pc.mean(), sd_pct=pc.std(ddof=1),
            mean_pct_div_sd=pc.mean()/pc.std(ddof=1),
            t=test.statistic, df=n-1, p=test.pvalue,
            pct_increased=100*(d>0).mean()))
        tag = f'{start}_to_{end}'
        cohort[f'{m}_change_{tag}'] = d
        cohort[f'{m}_pct_change_{tag}'] = pc
        if start == 'T1':
            cohort[f'{m}_change'] = d
            cohort[f'{m}_pct_change'] = pc
summary = pd.DataFrame(rows)
#All six paired t-tests shown here.
#The code below applies Holm's method to adjust p-values for multiple comparisons across the six paired t-tests.
pvals = summary.p.to_numpy()
order = np.argsort(pvals)
adjusted = np.minimum(1, np.maximum.accumulate((len(pvals)-np.arange(len(pvals)))*pvals[order]))
holm = np.empty(len(pvals)); holm[order] = adjusted
summary['p_holm_six_tests'] = holm
summary.to_csv(OUT / 'paired_test_results.csv', index=False)
cohort.to_csv(OUT / 'shared_cohort_changes.csv', index=False)
for m in COLS:
    sub = summary[summary.measure == m].reset_index(drop=True)
    fig, (ax, box) = plt.subplots(1, 2, figsize=(13, 5), gridspec_kw={'width_ratios':[1.1, 1.2]})
    x = sub.mean_change.to_numpy()
    ax.errorbar(x, [2, 1, 0], xerr=np.vstack([x-sub.ci_low, sub.ci_high-x]), fmt='o', capsize=7, color='steelblue')
    ax.axvline(0, color='gray', linestyle='--')
    ax.set(yticks=[2, 1, 0], yticklabels=['Baseline to T1', 'T1 to T2', 'Baseline to T2'], ylim=(-.6,2.6),
           xlabel=f'Mean change ({UNITS[m]})', title=f'{m}: mean change and 95% CI')
    lines = []
    for _, r in sub.iterrows():
        lines.append(f'{r.interval}\nn = {r.n}; t({r.df}) = {r.t:.3f}\n'
            f'Mean change = {r.mean_change:.3f}\n95% CI [{r.ci_low:.3f}, {r.ci_high:.3f}]\n'
            f'p = {pf(r.p)}; Holm p = {pf(r.p_holm_six_tests)}')
    note(box, '\n\n'.join(lines)+'\n\nPaired t-tests; Holm family = 6 tests.\nAll comparisons are exploratory.')
    fig.tight_layout()
    save(fig, f'02_{m}_paired_changes', 'Mean absolute changes across all three intervals with 95% t confidence intervals for the shared cohort.' \
    ' Positive values indicate increasing FF or longer TUG. Tests use interpolated values without propagating interpolation uncertainty.')


#6. Regression charts: equation, r, R-squared, p, slope CI, Spearman
#Below is a function to create regression plots for the relationship between FF and TUG at each visit, 
# as well as for changes in FF and TUG across intervals. It calculates both Pearson and Spearman correlations, 
# fits a linear regression line, and displays relevant statistics.
regrows = []
def regression_plot(x, y, title, xlabel, ylabel, name):
    pair = pd.DataFrame({'x':np.asarray(x),'y':np.asarray(y)}).dropna()
    x,y = pair.x.to_numpy(),pair.y.to_numpy()
    if len(x)<3 or np.ptp(x)==0 or np.ptp(y)==0:
        print('Skipping regression with insufficient variation:', name); return
    fit = stats.linregress(x,y)
    rho = stats.spearmanr(x,y)
    crit = stats.t.ppf(.975,len(x)-2)
    slo,shi = fit.slope-crit*fit.stderr,fit.slope+crit*fit.stderr
    grid = np.linspace(x.min(),x.max(),150)
    fitted = fit.intercept+fit.slope*grid
    residual_sd = np.sqrt(np.sum((y-fit.intercept-fit.slope*x)**2)/(len(x)-2))
    se = residual_sd*np.sqrt(1/len(x)+(grid-x.mean())**2/np.sum((x-x.mean())**2))
    fig,(ax,box)=plt.subplots(1,2,figsize=(12,5),gridspec_kw={'width_ratios':[1.6,1]})
    ax.scatter(x,y,alpha=.8)
    ax.plot(grid,fitted,color='firebrick',label='OLS fit')
    ax.fill_between(grid,fitted-crit*se,fitted+crit*se,color='firebrick',alpha=.13,label='95% CI for mean response')
    ax.set(xlabel=xlabel,ylabel=ylabel,title=title)
    ax.legend(fontsize=9)
    note(box,f'n = {len(x)}\ny = {fit.slope:.4f}x {fit.intercept:+.4f}\n'
        f'Pearson r = {fit.rvalue:.3f}\nR-squared = {fit.rvalue**2:.3f}\n'
        f'Slope p = {pf(fit.pvalue)}\nSlope 95% CI:\n[{slo:.4f}, {shi:.4f}]\n\n'
        f'Spearman rho = {rho.statistic:.3f}\nSpearman p = {pf(rho.pvalue)}\n\n'
        'Slope p also tests Pearson r = 0.\nUnadjusted exploratory p-values.')
    fig.tight_layout()
    save(fig,name,'Each point represents one subject. Ordinary least-squares line with a pointwise 95% confidence band for the mean response,' \
    ' not a prediction interval. Equation uses the units on the axes.')
    regrows.append(dict(analysis=name,n=len(x),slope=fit.slope,intercept=fit.intercept,
        pearson_r=fit.rvalue,r_squared=fit.rvalue**2,p=fit.pvalue,
        slope_ci_low=slo,slope_ci_high=shi,spearman_rho=rho.statistic,spearman_p=rho.pvalue))
for label,month in zip(LABELS,TARGETS):
    regression_plot(cohort[f'FF_{label}'],cohort[f'TUG_{label}'],f'FF and TUG at {month:g} months',AXLABEL['FF'],AXLABEL['TUG'],f'04_regression_{label}')

# Question 3: compare within-subject FF changes against TUG changes for ALL intervals.
for start, end in [('Baseline', 'T1'), ('T1', 'T2'), ('Baseline', 'T2')]:
    tag = f'{start}_to_{end}'
    regression_plot(cohort[f'FF_change_{tag}'], cohort[f'TUG_change_{tag}'],
                    f'FF change versus TUG change: {start} to {end}',
                    'Right-leg FF change (percentage points)', 'TUG change (s)',
                    f'05_change_regression_{tag}')

pd.DataFrame(regrows).to_csv(OUT/'regression_results.csv',index=False)


#Complete-visit sensitivity and descriptive individual correlations
#The code below identifies subjects with complete data across all three visits, 
# performs paired t-tests for each measure, and calculates descriptive correlations for individual subjects.
complete_ids=df.dropna(subset=TC+COLS['FF']+COLS['TUG']).subject_id
complete=cohort[cohort.subject_id.isin(complete_ids)]
sensitivity=[]
for m in COLS:
    test=stats.ttest_rel(complete[f'{m}_T2'],complete[f'{m}_T1'])
    sensitivity.append({'measure':m,'n':len(complete),'t':test.statistic,'p':test.pvalue})
pd.DataFrame(sensitivity).to_csv(OUT/'complete_visit_sensitivity.csv',index=False)
individual=[]
for _,r in df.iterrows():
    x=r[COLS['FF']].to_numpy(float);y=r[COLS['TUG']].to_numpy(float);t=r[TC].to_numpy(float)
    valid=np.isfinite(x)&np.isfinite(y)&np.isfinite(t)
    rr=np.nan
    if valid.sum()==3 and np.ptp(x[valid])>0 and np.ptp(y[valid])>0:
        rr=stats.pearsonr(x[valid],y[valid]).statistic
    individual.append({'subject_id':r.subject_id,'original_pairs':valid.sum(),'descriptive_r':rr})
pd.DataFrame(individual).to_csv(OUT/'individual_descriptive_correlations.csv',index=False)
```

## Conclusions:
**FF: mean change and 95% CI**

The fat fraction change was statistically significant from 0 to 22.13 months but not 11.15 to 22.13. We used a paired T-test and found the p value to be 0.0077.
From Baseline to T1, FF increased by approximately 0.74 percentage points. The unadjusted p-value is below 0.05, suggesting evidence of an increase. However, after Holm correction, the p-value becomes 0.0673, so this comparison is not statistically significant at the adjusted 0.05 threshold. From T1 to T2, FF increased by another 0.73 percentage points on average, but the confidence interval includes zero. There is insufficient statistical evidence of a change during this interval. From baseline to T2, FF increased by 1.47 percentage points, with an adjusted p-value of 0.0463. This is the only FF interval that remains statistically significant after correcting for the six paired comparisons. 


**TUG: mean change and 95% CI**
The TUG time Baseline to T1 TUG increased by approximately 0.10 seconds. This is a small average increase in TUG time, but it is not statistically significant with a Holm p-value of 0.84.
The T1 to T2 TUG decreased by approximately 0.28 seconds. Although this represents faster average performance, the difference is not statistically significant.
The baseline to T2 TUG decreased by approximately 0.18 seconds over the entire follow-up. Again, the change is not statistically significant.
All three confidence intervals include zero.

FF increased significantly across the full follow-up, while TUG did not demonstrate a significant average change.
